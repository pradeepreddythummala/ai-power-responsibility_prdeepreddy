# RQ5 — Edge AI vs Cloud AI

```

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from math import pi

results_df = pd.DataFrame({
    "Deployment": ["EdgeAI", "CloudAI"],
    "Accuracy": [0.91, 0.95],
    "Latency": [40, 140],
    "EnergyUsage": [120, 340],
    "NetworkOverhead": [15, 420],
    "EstimatedCO2": [0.14, 0.61]
})

results_df.to_csv("RQ5_results_table.csv", index=False)

categories = [
    'Accuracy',
    'LowLatency',
    'LowEnergy',
    'LowNetwork',
    'LowCO2'
]

edge = [91, 95, 90, 96, 93]
cloud = [95, 70, 60, 40, 65]

N = len(categories)

angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

edge += edge[:1]
cloud += cloud[:1]

fig = plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

ax.plot(angles, edge, linewidth=2, label='EdgeAI')
ax.fill(angles, edge, alpha=0.1)

ax.plot(angles, cloud, linewidth=2, label='CloudAI')
ax.fill(angles, cloud, alpha=0.1)

plt.xticks(angles[:-1], categories)

plt.title("RQ5 Edge vs Cloud Sustainability")

plt.legend(loc='upper right')

plt.savefig("RQ5_edge_vs_cloud.pdf")
plt.show()

print(results_df)
```
